# Data Processing

## Remove & handel missing value 

```python
df["Income"]=df["Income"].fillna(df["Income"].median())
df.isnull().sum()
```

```python
df.head()
```

## Feature engineering

```python
#year birth
df["age"]=2026-df["Year_Birth"]

#custmer joing date
df["Dt_Customer"]=pd.to_datetime(df["Dt_Customer"],dayfirst=True)

refarence_date=df["Dt_Customer"].max()
df["custormer_tenure_days"]=(refarence_date-df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
df.columns
```

```python
#spending 
df["total_spending"]=df["MntWines"]+df["MntFruits"]+df["MntMeatProducts"]+df["MntFishProducts"]+df["MntSweetProducts"]+df["MntGoldProds"]
```

```python
df.head()
```

```python
#children

df["total_children"]=df["Kidhome"]+df["Teenhome"]
df.head()
```

```python
#Education

df["Education"].value_counts()
```

```python
df["Education"]=df["Education"].replace({
    "basic":"undergraduate",
    "2n Cycle":"undergraduate",
    "Graduation":"graduate",
    "PhD":"postgraduate",
    "Master":"postgraduate"
})

df.head()
```

```python
# Marital status

df["Marital_Status"].value_counts()
```

```python
df["living_with"]=df["Marital_Status"].replace({
    "Married":"partner",
    "Together":"partner",
    "single":"Alone",
    "Divorced":"Alone",
    "Widow":"Alone",
    "Absurd":"Alone",
    "YOLO":"Alone"
})
df.head()
```

```python
df["living_with"].value_counts()
```

## Scaling

```python
from sklearn.preprocessing import StandardScaler
X=df_encoded
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
X_scaled
```

# DATA VISULIZATION

## 2D 

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
X_pca=pca.fit_transform(X_scaled)

plt.scatter(X_pca[:,0],X_pca[:,1])
pca.explained_variance_ratio_
```

## 3D

```python
pca=PCA(n_components=3)
X_pca=pca.fit_transform(X_scaled)

fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2])
ax.view_init(elev=10, azim=130) 

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.set_title("3D_projection")
```

# ANALYZE K VALUE

## Elbow Method

```python
pip install kneed 
```

```python

from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)  
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)

knee = KneeLocator(
    range(1, 11),
    wcss,
    curve="convex",
    direction="decreasing"
)

optimal_k = knee.elbow

print("\nBEST K= ", optimal_k,"\n")

plt.plot(range(1, 11), wcss, marker="o")
plt.xlabel("k")
plt.ylabel("wcss")
plt.show()  
```

# Silhouett Score

```python
from sklearn.metrics import silhouette_score
scores=[]
for k in range(2,11):
    kmeans = KMeans(n_clusters=k, random_state=42)  
    labels=kmeans.fit_predict(X_pca)
    score=silhouette_score(X_pca,labels)
    scores.append(score)

plt.plot(range(2, 11), scores, marker="o")
plt.xlabel("k")
plt.ylabel("silhouette_score")
plt.show()  
```

## combination of both plot

```python
k_range = range(2, 11)

fig, ax1 = plt.subplots(figsize=(8, 6))

ax1.plot(
    k_range,
    wcss[1:],              
    marker="o",
    color="blue"
)

ax1.set_xlabel("k")
ax1.set_ylabel("wcss")

ax2 = ax1.twinx()          

ax2.plot(
    k_range,
    scores[:len(k_range)],
    marker="x",
    color="red",
    linestyle="--"
)

ax2.set_ylabel("ss")

plt.show()
```

# CLUSTERING 

```python
kmean = KMeans(n_clusters=4, random_state=42)
labels_kmeans = kmean.fit_predict(X_pca)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    X_pca[:, 2],
    c=labels_kmeans
)

ax.view_init(elev=10, azim=130)   

plt.show()
```

# AGGLOMERATIVE CLUSTERING

```python
from sklearn.cluster import AgglomerativeClustering

agg_clf = AgglomerativeClustering(n_clusters=4, linkage="ward")
labels_agg = agg_clf.fit_predict(X_pca)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    X_pca[:, 2],
    c=labels_agg
)   

ax.view_init(elev=20, azim=130)   

plt.show()
```
